#### - **A theoretical approach** uses abstraction and assumptions to understand, explain, and predict real-world phenomena.

```python
from sklearn.decomposition import PCA

img_gray = Image.open(urlopen(req)).convert('L')
img_np = np.array(img_gray)
print(img_np.shape)

components_list = [5, 10, 15, 30]

fig, axes = plt.subplots(1, 5)

axes[0].imshow(img_np, cmap='gray')
axes[0].set_title("Original (Full)")
axes[0].axis('off')

for i, n_comp in enumerate(components_list, start=1):
    pca = PCA(n_components=n_comp)
    img_transformed = pca.fit_transform(img_np)
    print(img_transformed.shape)
    img_inverted = pca.inverse_transform(img_transformed)
    axes[i].imshow(img_inverted, cmap='gray')
    
    axes[i].set_title(f"PC = {n_comp}")
    axes[i].axis('off')
    
plt.tight_layout()
plt.show()
```

## Why divide by `N` or `n − 1`?

**Question:** Why do we divide population variance by the number of observations, but sample variance by one fewer?

**Step A — The whole population.** Suppose the **entire population** consists of five simplified GPA-like scores: `1, 2, 3, 4, 5`. We know everyone's score, so we know the **true population mean**. Population variance is simply the *average squared distance* from that mean:

$$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}.$$

Here `N = 5`. There is nothing to estimate about this particular five-person population.

```python
all_scores = np.array([1, 2, 3, 4, 5])

people = len(all_scores)
true_mean = all_scores.mean()
squared_distances = (all_scores - true_mean) ** 2
true_variance = squared_distances.sum() / people

print("Entire population:", all_scores)
print("Population mean:", true_mean)
print("Squared distances:", squared_distances)
print("Total squared distance:", squared_distances.sum())
print("Population variance = 10 / 5 =", true_variance)
```

**Step B — Observe only a sample.** Now imagine you see only three scores: `2, 3, 5`. In real inference you do **not** know the population mean, so you calculate the mean of these three observed scores.

Notice what happens to the **deviations** (score minus sample mean): they **must add up to zero**. Once you know the first two deviations, the third has no choice. We say the three deviations have only **two degrees of freedom**.

$$s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1}.$$

**Important:** Deviations from the population mean also sum to zero when calculated over the full population. The deeper reason for the different denominator is that the **sample mean is estimated from the same data**. This makes the squared deviations too small *on average* when used to estimate population variance. Dividing by `n − 1` corrects that average downward bias for independent, identically distributed sampling.

---

```python
three_scores = np.array([2, 3, 5])
n_three = len(three_scores)
mean_three = three_scores.mean()
three_deviations = three_scores - mean_three
sum_squares = (three_deviations ** 2).sum()

print("Observed sample:", three_scores)
print("Sample mean:", round(mean_three, 3))
print("Deviations:", np.round(three_deviations, 3))
print("Sum of deviations:", round(three_deviations.sum(), 10))
print("First + second + third deviation = 0")
print("First two determine the third -> degrees of freedom =", n_three - 1)
print("Variance using n:", round(sum_squares / n_three, 3))
print("Variance using n - 1:", round(sum_squares / (n_three - 1), 3))
print("True population variance (known only in our simulation):", true_variance)
```

```python
rng_variance = np.random.RandomState(123)
practice_samples = rng_variance.choice(all_scores, size=(10000, 3), replace=True)

variance_using_n = practice_samples.var(axis=1, ddof=0)
variance_using_n_minus_1 = practice_samples.var(axis=1, ddof=1)

print("Actual population variance:", round(true_variance, 3))
print("Average estimate using n:", round(variance_using_n.mean(), 3))
print("Average estimate using n - 1:", round(variance_using_n_minus_1.mean(), 3))

plt.bar(["True population", "Divide by n", "Divide by n - 1"],
        [true_variance, variance_using_n.mean(), variance_using_n_minus_1.mean()],
        color=["black", "0.75", "0.45"], edgecolor="black")
plt.ylabel("Variance (squared score units)")
plt.title("Average of 10,000 variance estimates (sample size = 3)")
plt.tight_layout()
plt.show()
```

## Standard Deviation and Standard Error: two different kinds of spread

| Measure | What does it describe? | Unit |
|---|---|---|
| SD | Differences between individual values | GPA points |
| SE of the mean | Differences between means from repeated samples | GPA points |
| Variance | Squared differences from the mean | GPA points squared |

With independent draws and a finite variance, population SE is sigma divided by the square root of n. Sigma is the population SD. We usually do not know sigma, so we estimate it with the sample SD, s.

$$
SE(\bar{x})=\frac{\sigma}{\sqrt{n}},\qquad
\widehat{SE}(\bar{x})=\frac{s}{\sqrt{n}}.
$$

SE tells us the typical size of sampling variation. It does **not** tell us the exact error in our one sample. A larger n usually makes SE smaller under these sampling rules.

# Central Limit Theorem (CLT): Why It Saves Our Lives

### Central Limit Theorem

If we repeatedly take independent random samples of size n
from a population with a finite mean μ and standard deviation σ,
the sampling distribution of the **sample mean** becomes
approximately normal (bell-shaped) as n increases,
regardless of the original population's distribution.

### Three Key Properties

1. **Shape:** The sampling distribution approaches a normal
   distribution as n increases (n ≥ 30 is a common rule of thumb).

2. **Center:** The mean of the sampling distribution equals
   the population mean (μ).

3. **Spread:** The standard error is σ / √n.
   Larger samples provide more precise estimates of the
   population mean.

### Why Is the CLT So Important?

**One of the Most Important Statistical Foundations of Social Science:
From Social Phenomena to Scientific Inference**

- Makes it possible to draw statistical inferences about
  populations using sample data.
- Provides a foundation for confidence intervals and
  hypothesis testing.
- Enables scientific analysis of complex social and economic
  phenomena in business, economics, sociology, and political science.
- From Social Events to Social Sciences

```python
REPETITIONS = 1000
samples = rng.choice(population_gpa, size=(REPETITIONS, N), replace=True)
sample_means = samples.mean(axis=1)
sample_sds = samples.std(axis=1, ddof=1)
theoretical_se = np.std(population_gpa, ddof=0) / np.sqrt(N)
empirical_se = sample_means.std(ddof=1)
print(f"Mean of sample means={sample_means.mean():.4f}")
print(f"Empirical SE={empirical_se:.4f}; population SE={theoretical_se:.4f}")

fig, axes = plt.subplots(1, 2)
axes[0].hist(population_gpa, bins=30, color="0.75", edgecolor="black")
axes[0].set(title="Population", xlabel="GPA", ylabel="Records")
axes[1].hist(sample_means, bins=30, color="0.75", edgecolor="black")
axes[1].axvline(population_mean, color="black", linestyle="--")
axes[1].set(title=f"Means of {N} draws", xlabel="Sample mean GPA", ylabel="Samples")
```

## Inferential Statistics: From a Sample to a Population 


We know the GPAs of **30 students**. Can we use these 30 students to learn something about **all 3,000 students**?

- **Population:** all 3,000 students (the whole group we care about).
- **Sample:** 30 students we actually observe.
- **Sample mean** ($\bar{x}$): the average GPA of those 30 students.
- **Population mean** ($\mu$): the average GPA of all 3,000 students.
- **Inferential statistics:** using the sample to say something about the population, while allowing for uncertainty.

- A sample mean is not always equal to the population mean. Different samples give different answers.
- We created the 3,000-student population on a computer, so we can see the true population mean and standard deviation. 
- In real research, we usually **do not know** those true values. We will first learn with the known values, then learn what to do without them.

```python
from scipy import stats

z_example = round(demo_z, 2)   # -0.76

area_left = stats.norm.cdf(z_example)
area_right = 1 - area_left

print(f"Z-score: {z_example:.2f}")
print(f"Left of z={z_example:.2f}:  {area_left:.4f} = {area_left:.2%}")
print(f"Right of z={z_example:.2f}: {area_right:.4f} = {area_right:.2%}")
```

### Review: From Sample to Population

| Step | What we do | Why it matters |
|---|---|---|
| **1. Sample** | Pick 30 students | We cannot usually measure everyone |
| **2. Mean** | Find $\bar{x}=2.934$ | This is our sample estimate |
| **3. SE** | Measure how sample means vary | We need to know our uncertainty |
| **4. Z** | Measure distance in true SE units | Teaches us the standard normal scale, **when $\mu$ and $\sigma$ are known** |
| **5. Z-table** | Read the area to the left of a Z-score | Describes the chance of possible sample results under a given model |
| **6. t-test** | Test $H_0:\mu=3.0$ using sample SD | Works when population SD is unknown, with appropriate assumptions |
| **7. 95% CI** | Get a range for the unknown mean | Shows how uncertain our estimate is |

The existing `EIU_population` already contains `exam_total` and `gpa`. We do **not** replace those values.

To make a classroom example with a pattern we can study, we create an IELTS score related to GPA and entrance-exam score **plus random noise**. The invented relationship is chosen for practice; **it is not a claim about real students**.

We make five additional columns: `student_name`, `ielts_exam_date`, `ielts_study_months`, `ielts_score`, and `ielts_pass`.

- `ielts_score`: rounded to the nearest **0.5 band**, like actual IELTS scores.

```python
from faker import Faker
from datetime import date

fake = Faker()
Faker.seed(2026)
rng_ielts = np.random.default_rng(2026)

EIU_population = EIU_population.copy()
student_count = len(EIU_population)

EIU_population["student_name"] = [fake.name() for _ in range(student_count)]
EIU_population["ielts_exam_date"] = [
    fake.date_between(start_date=date(2024, 1, 1), end_date=date(2026, 10, 31))
    for _ in range(student_count)
]
EIU_population["ielts_study_months"] = rng_ielts.integers(4, 21, student_count)

raw_score = (5.90
             + 0.55 * (EIU_population["gpa"] - 3.0)
             + 0.16 * (EIU_population["exam_total"] - 28.0)
             + rng_ielts.normal(0, 0.65, student_count))

EIU_population["ielts_score"] = np.clip(np.round(raw_score * 2) / 2, 0, 9)
EIU_population["ielts_pass"] = (EIU_population["ielts_score"] > 6.0).astype(int)

EIU_population["ielts_pass_within_1year"] = (
    (EIU_population["ielts_score"] > 6.0)
    & (EIU_population["ielts_study_months"] <= 12)
).astype(int)

print("Invented population size:", len(EIU_population))
print("IELTS pass rate (score > 6):", f'{EIU_population["ielts_pass"].mean():.1%}')
display(EIU_population[["student_name", "exam_total", "gpa",
                        "ielts_score", "ielts_pass",
                        "ielts_study_months", "ielts_pass_within_1year"]].head())
```

## What question does linear regression answer?

Each point in the plot is a sampled student. A straight line can summarize the overall *association*. It will not predict every student's GPA perfectly.

## Fit a line using our sample


$$\widehat{GPA}=b_0+b_1\times exam\_total$$

- **$b_0$ (intercept):** the predicted GPA when `exam_total = 0`. That may be outside the useful exam-score range.
- **$b_1$ (slope):** how much the predicted GPA changes for **one more entrance-exam point**.
- **$\widehat{GPA}$:** the model's *predicted* GPA, not the actual GPA.


## Interpret the regression slope

A **positive** slope means a student with one more entrance-exam point is predicted to have a slightly **higher GPA on average**. A **negative** slope means a slightly lower GPA.

The slope is in units of **GPA points per exam point**. For a **5-point** difference in exam score, multiply the estimated slope by **5**. These are *predicted differences*, not guaranteed changes for any student.

**Association is not causation:** a higher entrance-exam score may be associated with higher GPA, but this analysis cannot show that increasing the exam score **causes** GPA to rise.

# Logistic Regression (Logit): GPA + Exam Total → IELTS Pass (0/1)

## Why is this classification and not linear regression?

Here the outcome is not a GPA number. It is **yes/no**:

- **Y = `ielts_pass` = 1** if invented IELTS score **> 6.0**.
- **Y = `ielts_pass` = 0** if invented IELTS score **≤ 6.0**.

The predictors are **X = GPA and entrance-exam total score**. Do **not** put `ielts_score` into the predictors; it directly tells us the answer and would cause **data leakage**. We also do not include `ielts_pass_within_1year`.

**Logistic regression (Logit)** predicts a probability between **0% and 100%**. It is a regression *model* used for **binary classification**.


## Convert log-odds coefficients into odds ratios

Logit coefficients are hard to understand directly. We can calculate **odds ratio = $e^b$**.

- **OR = 1:** no difference in odds associated with one more unit of the predictor.
- **OR > 1:** higher odds of IELTS pass.
- **OR < 1:** lower odds of IELTS pass.
