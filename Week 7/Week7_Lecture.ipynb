#### **A theory is a simplified representation of reality.**

#### - **A theoretical approach** uses abstraction and assumptions to understand, explain, and predict real-world phenomena.
#### - **A practical approach** focuses on solving real-world problems and achieving specific objectives under conditions and constraints.
#### - **Theory and practice are complementary, not opposing approaches.**

#### Theory guides practice, while practice tests, informs, and refines theory.

#### **We need both:** Theory helps us understand how the world works. Practice helps us solve problems in the real world.

```python
# Cosine Similarity: normalized dot product by norm
"""
a*b = |a||b|cos(theta)
cos(theta) = a*b / (|a||b|)
"""
D = E
sim = []
for i in range(len(E)):
    q_v = E[i]
    print(f"Query: {words[i]}")
    for j in range(len(D)):
        d_v = D[j]

        norm_q = np.linalg.norm(q_v)
        norm_d = np.linalg.norm(d_v)
        
        cos_sim = np.dot(q_v, d_v) / (norm_q * norm_d)

        sim.append(cos_sim)

        print(cos_sim)
```

## Why divide by `N` or `n − 1`?

**Question:** Why do we divide population variance by the number of observations, but sample variance by one fewer?

**Step A — The whole population.** Suppose the **entire population** consists of five simplified GPA-like scores: `1, 2, 3, 4, 5`. We know everyone's score, so we know the **true population mean**. Population variance is simply the *average squared distance* from that mean:

$$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}.$$

Here `N = 5`. There is nothing to estimate about this particular five-person population.

```python
all_scores = np.array([1, 2, 3, 4, 5])

people = len(all_scores)
true_mean = all_scores.mean()
squared_distances = (all_scores - true_mean) ** 2
true_variance = squared_distances.sum() / people

print("Entire population:", all_scores)
print("Population mean:", true_mean)
print("Squared distances:", squared_distances)
print("Total squared distance:", squared_distances.sum())
print("Population variance = 10 / 5 =", true_variance)
```

**Step B — Observe only a sample.** Now imagine you see only three scores: `2, 3, 5`. In real inference you do **not** know the population mean, so you calculate the mean of these three observed scores.

Notice what happens to the **deviations** (score minus sample mean): they **must add up to zero**. Once you know the first two deviations, the third has no choice. We say the three deviations have only **two degrees of freedom**.

$$s^2 = \frac{\sum (x_i - \bar{x})^2}{n-1}.$$

**Important:** Deviations from the population mean also sum to zero when calculated over the full population. The deeper reason for the different denominator is that the **sample mean is estimated from the same data**. This makes the squared deviations too small *on average* when used to estimate population variance. Dividing by `n − 1` corrects that average downward bias for independent, identically distributed sampling.

---

```python
three_scores = np.array([2, 3, 5])
n_three = len(three_scores)
mean_three = three_scores.mean()
three_deviations = three_scores - mean_three
sum_squares = (three_deviations ** 2).sum()

print("Observed sample:", three_scores)
print("Sample mean:", round(mean_three, 3))
print("Deviations:", np.round(three_deviations, 3))
print("Sum of deviations:", round(three_deviations.sum(), 10))
print("First + second + third deviation = 0")
print("First two determine the third -> degrees of freedom =", n_three - 1)
print("Variance using n:", round(sum_squares / n_three, 3))
print("Variance using n - 1:", round(sum_squares / (n_three - 1), 3))
print("True population variance (known only in our simulation):", true_variance)
```

```python
rng_variance = np.random.RandomState(123)
practice_samples = rng_variance.choice(all_scores, size=(10000, 3), replace=True)

variance_using_n = practice_samples.var(axis=1, ddof=0)
variance_using_n_minus_1 = practice_samples.var(axis=1, ddof=1)

print("Actual population variance:", round(true_variance, 3))
print("Average estimate using n:", round(variance_using_n.mean(), 3))
print("Average estimate using n - 1:", round(variance_using_n_minus_1.mean(), 3))

plt.bar(["True population", "Divide by n", "Divide by n - 1"],
        [true_variance, variance_using_n.mean(), variance_using_n_minus_1.mean()],
        color=["black", "0.75", "0.45"], edgecolor="black")
plt.ylabel("Variance (squared score units)")
plt.title("Average of 10,000 variance estimates (sample size = 3)")
plt.tight_layout()
plt.show()
```

## Standard Deviation and Standard Error: two different kinds of spread

| Measure | What does it describe? | Unit |
|---|---|---|
| SD | Differences between individual values | GPA points |
| SE of the mean | Differences between means from repeated samples | GPA points |
| Variance | Squared differences from the mean | GPA points squared |

With independent draws and a finite variance, population SE is sigma divided by the square root of n. Sigma is the population SD. We usually do not know sigma, so we estimate it with the sample SD, s.

$$
SE(\bar{x})=\frac{\sigma}{\sqrt{n}},\qquad
\widehat{SE}(\bar{x})=\frac{s}{\sqrt{n}}.
$$

SE tells us the typical size of sampling variation. It does **not** tell us the exact error in our one sample. A larger n usually makes SE smaller under these sampling rules.

# Central Limit Theorem (CLT): Why It Saves Our Lives

### Central Limit Theorem

If we repeatedly take independent random samples of size n
from a population with a finite mean μ and standard deviation σ,
the sampling distribution of the **sample mean** becomes
approximately normal (bell-shaped) as n increases,
regardless of the original population's distribution.

### Three Key Properties

1. **Shape:** The sampling distribution approaches a normal
   distribution as n increases (n ≥ 30 is a common rule of thumb).

2. **Center:** The mean of the sampling distribution equals
   the population mean (μ).

3. **Spread:** The standard error is σ / √n.
   Larger samples provide more precise estimates of the
   population mean.

### Why Is the CLT So Important?

**One of the Most Important Statistical Foundations of Social Science:
From Social Phenomena to Scientific Inference**

- Makes it possible to draw statistical inferences about
  populations using sample data.
- Provides a foundation for confidence intervals and
  hypothesis testing.
- Enables scientific analysis of complex social and economic
  phenomena in business, economics, sociology, and political science.
- From Social Events to Social Sciences

## Inferential Statistics: From a Sample to a Population 


We know the GPAs of **30 students**. Can we use these 30 students to learn something about **all 3,000 students**?

- **Population:** all 3,000 students (the whole group we care about).
- **Sample:** 30 students we actually observe.
- **Sample mean** ($\bar{x}$): the average GPA of those 30 students.
- **Population mean** ($\mu$): the average GPA of all 3,000 students.
- **Inferential statistics:** using the sample to say something about the population, while allowing for uncertainty.

- A sample mean is not always equal to the population mean. Different samples give different answers.
- We created the 3,000-student population on a computer, so we can see the true population mean and standard deviation. 
- In real research, we usually **do not know** those true values. We will first learn with the known values, then learn what to do without them.

### Step 1. Start with the 30 students we already sampled

30 GPAs in `sample_gpa`.

The average of these 30 GPAs is about **2.934**. The average of all 3,000 simulated GPAs is about **2.999**.


### Step 2. How far is our sample mean from the true mean?

Subtract the population mean from the sample mean:

$$\text{Difference}=\bar{x}-\mu$$

In our example, $2.934 - 2.999 \approx -0.065$.

The minus sign simply means **our sample average is lower** than the population average.

But is $-0.065$ a **big** difference? We cannot know just from this number. We need to know how much sample means usually move around when we take new samples.

### Step 3. Standard Error (SE): How much do sample means change?

Imagine that we repeat this experiment:

1. Pick 30 students at random. Find their average GPA.
2. Pick another 30 students. Find that average GPA.
3. Do this many times.

We will get many different **sample averages**. The standard deviation of those averages is the **standard error (SE)**.

**SD versus SE:**

| Name | What changes? | What does it tell us? |
|---|---|---|
| **SD** | Individual students' GPAs | How different individual students are |
| **SE** | Averages of different samples | How much a sample average usually changes |

If we know the **population SD** ($\sigma$), the basic formula for independent draws is:

$$SE_{\mathrm{true}}=\frac{\sigma}{\sqrt{n}}$$

Here $n=30$. For this computer-made population, $\sigma \approx 0.470$.


### Step 4. Convert the difference to a Z-score

The raw difference ($-0.065$) is hard to judge. So let's measure it in units of **SE** instead of GPA points.

$$z=\frac{\text{sample mean} - \text{population mean}}{SE_{\mathrm{true}}}
=\frac{\bar{x}-\mu}{\sigma/\sqrt{n}}$$

In everyday English: **Z-score = how many standard errors above or below the population mean the sample mean is.**

- $z=0$: the sample mean equals the population mean.
- $z=+1$: the sample mean is one SE **above** the population mean.
- $z=-1$: the sample mean is one SE **below** the population mean.


### Step 5. Why can we use a Z-table?

The **Central Limit Theorem** says that, under suitable conditions, the distribution of sample means becomes **approximately normal** as the sample size gets large.

When we standardize those sample means, their distribution is approximately the **standard normal distribution**:

$$Z=\frac{\bar{X}-\mu}{\sigma/\sqrt{n}}\ \approx\ N(0,1)$$

This distribution has mean 0 and standard deviation 1.

- The **Z-score** tells us where our sample mean sits on the horizontal axis.
- The **Z-table** tells us the **area to the left** of that Z-score.
- That area is a **probability** across possible samples.


### Step 6. Read the Z-table: What does Z = −0.76 mean?

In Step 4, we used our **actual 30-student GPA sample**:

| What we know | Value |
|---|---:|
| Population mean GPA ($\mu$) | 2.999 |
| Our sample mean GPA ($\bar{x}$) | 2.934 |
| Standard error (SE) | 0.0859 |
| Our Z-score ($z$) | **−0.76** |

**What does −0.76 mean?** Our sample average (2.934) is **0.76 standard errors below** the population average (2.999).

**1. Look up −0.76 in the Z-table**

A [Z-table](https://www.simplypsychology.org/z-table.html) gives the area **to the left** of a Z-score:

$$P(Z\leq -0.76)\approx 0.2236 = 22.36\%$$

Read this as: **"About 22.36% of possible sample averages are as low as ours or lower."**

- **22.36%** of possible sample averages are about **2.934 or lower**.
- **77.64%** are **higher than 2.934**.

**2. Imagine taking 1,000 new samples**

Each time, pick 30 students and calculate their average GPA. If the same population model holds, we expect roughly **224 out of 1,000** sample averages to be this low or lower, and **776** to be higher.

This places our sample average near the **22nd percentile** of possible sample averages.


```python
from scipy import stats

z_example = round(demo_z, 2)   # -0.76

area_left = stats.norm.cdf(z_example)
area_right = 1 - area_left

print(f"Z-score: {z_example:.2f}")
print(f"Left of z={z_example:.2f}:  {area_left:.4f} = {area_left:.2%}")
print(f"Right of z={z_example:.2f}: {area_right:.4f} = {area_right:.2%}")
```

### Step 7. See the area on the bell curve

The shaded area below is the part of the **standard normal curve to the left of $z=-0.76$**.

- **Shaded area: about 22.36%.** That is what the Z-table reports.
- **Unshaded area: about 77.64%.** That is the probability to the right.
- The **dashed line** marks our Z-score, −0.76.

**Remember:** A Z-table gives the **area** under the curve, **not** the curve's height.

```python
z_values = np.linspace(-3.5, 3.5, 300)
heights = stats.norm.pdf(z_values)

plt.plot(z_values, heights, color="black")

plt.fill_between(z_values, heights, where=z_values <= z_example,
                 color="skyblue", alpha=0.5,
                 label=f"Left area = {area_left:.2%}")
plt.axvline(z_example, color="red", linestyle="--",
            label=f"z = {z_example:.2f}")

plt.title(f"Z-table: {area_left:.2%} is to the left of z = {z_example:.2f}")
plt.xlabel("Z-score")
plt.ylabel("Curve height (density)")
plt.legend()
plt.show()
```

### Step 8. Where does the Z-table number come from? (Optional math)

There are **two different things**:

- **PDF (probability density function):** tells us the *height* of the curve at a Z-score.
- **CDF (cumulative distribution function):** tells us the total *area to the left* of a Z-score. The Z-table gives CDF values.

The **standard normal distribution** has this PDF:

$$f(z)=\frac{1}{\sqrt{2\pi}}e^{-z^2/2}$$

Here $\pi$ and $e$ are fixed mathematical constants. This formula describes the **normal** distribution, not every distribution with mean 0 and SD 1.

To calculate the area to the left of **our $z=-0.76$**, we add up the tiny areas under the curve. In calculus, we write:

$$P(Z\leq -0.76)=\int_{-\infty}^{-0.76}f(z)\,dz\approx0.2236$$

You do **not** need to do the integral yourself. The Z-table or Python already calculates it.

**Remember: PDF = height; CDF/Z-table = accumulated area.**

```python
print(f"Curve height at z={z_example:.2f} (PDF): {stats.norm.pdf(z_example):.4f}")
print(f"Area left of z={z_example:.2f} (CDF):  {stats.norm.cdf(z_example):.4f}")
```

### Step 9. Why is 1.96 important for a 95% interval?

The standard normal curve is symmetric (the two sides mirror each other).

- **2.5%** of the area is below $z=-1.96$.
- **95%** of the area is between $-1.96$ and $+1.96$.
- **2.5%** of the area is above $z=+1.96$.

$$P(-1.96\leq Z\leq1.96)\approx0.95$$



### Step 10. What conclusion can we draw from our GPA sample?

We observed an average GPA of **2.934** among 30 students. The computer-generated population mean is **2.999**.

**First question: Is 2.934 below 2.999?** Yes. The difference is **−0.065 GPA points**. In standard-error units, we found **Z ≈ −0.76**.

**Second question: Is the difference unusual?** The Z-table gives a left-side area of about **22.36%**. If the population mean really is 2.999 and our normal approximation is good, roughly **224 out of 1,000** random samples of 30 students would have an average GPA **2.934 or lower**. This is not rare. Different random samples naturally give different averages.

**Third question: Can we reject the claim that the population mean is 2.999?** We need a *hypothesis test*. To test whether the mean is **different in either direction**, write:

- **Null hypothesis:** $H_0:\mu=2.999$.
- **Alternative hypothesis:** $H_1:\mu\ne2.999$.

The **two-sided p-value** counts sample results **at least as far away** from 2.999 as ours, on *both* sides:

$$p=P(|Z|\geq0.76)\approx2\times P(Z\leq-0.76)\approx0.4473.$$



```python

z_observed = -0.76
left_area = stats.norm.cdf(z_observed)
two_sided_p = 2 * stats.norm.sf(abs(z_observed))

print(f"Left-side Z-table area: {left_area:.2%}")
print(f"Two-sided p-value:      {two_sided_p:.2%}")
print("Decision at 5%: Do not reject H0" if two_sided_p > 0.05
      else "Decision at 5%: Reject H0")
```

#### Draw the two-sided p-value on the normal curve

The earlier picture shaded just the left-side **22.36%**. A test asking whether the mean is **different**, not just **lower**, counts extreme sample means on **both sides**. The two shaded regions together are the **44.73% p-value**.


```python
z_axis = np.linspace(-3.5, 3.5, 300)
curve = stats.norm.pdf(z_axis)

plt.figure(figsize=(8, 4))
plt.plot(z_axis, curve, color="black")
plt.fill_between(z_axis, curve, where=z_axis <= -0.76,
                 color="cornflowerblue", alpha=0.45, label="Left tail")
plt.fill_between(z_axis, curve, where=z_axis >= 0.76,
                 color="cornflowerblue", alpha=0.45, label="Right tail")
plt.axvline(-0.76, color="black", linestyle="--")
plt.axvline(0.76, color="black", linestyle="--")
plt.title(f"Two-sided p-value: {two_sided_p:.2%} (both tails)")
plt.xlabel("Z-score")
plt.ylabel("Probability density")
plt.legend()
plt.show()
```

### Step 11. In reality, We do NOT know the population mean or SD

So far we used the true population mean and SD because we created the entire population on the computer.

In a real study, we usually only have the 30 students. We cannot calculate our real Z-score using an unknown population mean and unknown population SD.

Instead, we can ask a **testable question**:

> **Could the population mean GPA be 3.0?**

This claim is the **null hypothesis**:

$$H_0:\mu=3.0$$

We know the sample mean. We estimate SE using the **sample SD** ($s$) instead of the population SD ($\sigma$):

$$SE_{\mathrm{estimated}}=\frac{s}{\sqrt n}$$

Then we calculate a **t-score**, not a Z-score:

$$t=\frac{\bar{x}-3.0}{s/\sqrt n}$$


```python
demo_sample_sd = sample_gpa_pd.std(ddof=1)
demo_se_est = demo_sample_sd / np.sqrt(demo_n)
demo_test_mean = 3.0
demo_t = (demo_sample_mean - demo_test_mean) / demo_se_est

print(f"Sample SD: {demo_sample_sd:.3f}")
print(f"Estimated SE: {demo_se_est:.4f}")
print(f"t-score for H0: mean = 3.0: {demo_t:.2f}")
```

### Step 12. Use the t-score to calculate a p-value

The **p-value** answers this question:

> "**If** the population mean really were 3.0, how unusual would our sample mean be (or a result even farther from 3.0)?"

For a **two-sided test** (the mean might be higher *or* lower), we count **both tails** of the t-distribution.

- A **small** p-value means our observed result would be unusual if the null hypothesis were true.
- A **large** p-value means this result is not unusual enough to reject that hypothesis.
- A p-value is **not** the probability that the hypothesis is true.

The degrees of freedom for this one-sample t-test are $n-1=29$.

```python
degrees_of_freedom = demo_n - 1
p_value = 2 * stats.t.sf(abs(demo_t), df=degrees_of_freedom)
print(f"Two-sided p-value: {p_value:.3f}")
```

### Step 13. Confidence interval: Give a reasonable range for the population mean

A confidence interval goes one step further than giving only a sample average.

For a mean, a common **95% t confidence interval** is:

$$\bar{x}\;\pm\;t_{0.975,\;n-1}\times\frac{s}{\sqrt n}$$

In easy English:

> **Sample mean ± (critical t-value × estimated SE)**

The range is meant to capture the **unknown population mean**, not 95% of individual students' GPAs.

If we repeated the sampling process many times and made one interval per sample, about **95% of those intervals** would include the true population mean, when the statistical assumptions hold.

```python
t_critical = stats.t.ppf(0.975, df=degrees_of_freedom)
margin = t_critical * demo_se_est
lower = demo_sample_mean - margin
upper = demo_sample_mean + margin

print(f"95% CI: ({lower:.3f}, {upper:.3f})")
```

| Step | What we do | Why it matters |
|---|---|---|
| **1. Sample** | Pick 30 students | We cannot usually measure everyone |
| **2. Mean** | Find $\bar{x}=2.934$ | This is our sample estimate |
| **3. SE** | Measure how sample means vary | We need to know our uncertainty |
| **4. Z** | Measure distance in true SE units | Teaches us the standard normal scale, **when $\mu$ and $\sigma$ are known** |
| **5. Z-table** | Read the area to the left of a Z-score | Describes the chance of possible sample results under a given model |
| **6. t-test** | Test $H_0:\mu=3.0$ using sample SD | Works when population SD is unknown, with appropriate assumptions |
| **7. 95% CI** | Get a range for the unknown mean | Shows how uncertain our estimate is |

# Linear Regression (OLS): `exam_total` → `GPA`

## What question does linear regression answer?

Our question: **Do students with higher entrance-exam total scores tend to have higher GPA?**

- **X = `exam_total`** (predictor, entrance-exam score)
- **Y = `gpa`** (outcome, GPA)

Each point in the plot is a sampled student. A straight line can summarize the overall *association*. It will not predict every student's GPA perfectly.


```python
# Plot
```

## Fit a line using our sample


$$\widehat{GPA}=b_0+b_1\times exam\_total$$

- **$b_0$ (intercept):** the predicted GPA when `exam_total = 0`. That may be outside the useful exam-score range.
- **$b_1$ (slope):** how much the predicted GPA changes for **one more entrance-exam point**.
- **$\widehat{GPA}$:** the model's *predicted* GPA, not the actual GPA.


## Interpret the regression slope

A **positive** slope means a student with one more entrance-exam point is predicted to have a slightly **higher GPA on average**. A **negative** slope means a slightly lower GPA.

The slope is in units of **GPA points per exam point**. For a **5-point** difference in exam score, multiply the estimated slope by **5**. These are *predicted differences*, not guaranteed changes for any student.

**Association is not causation:** a higher entrance-exam score may be associated with higher GPA, but this analysis cannot show that increasing the exam score **causes** GPA to rise.


## How do we infer a POPULATION relationship from a SAMPLE slope?

This is the **same idea** as sampling means. If we took another random sample of 240 students, our fitted slope would change.

- **Sample slope $b_1$:** our best estimate from these students.
- **Population slope $\beta_1$:** the relationship we want to learn about for all 3,000 students.
- **SE of the slope:** how much the slope estimate would vary across samples (approximately).
- **95% confidence interval:** a range of possible population slopes compatible with our data and model.
- **p-value:** how strongly the sample contradicts a proposed **zero population slope**.

This **SE** is *not* the earlier `SD / sqrt(n)` for a sample mean. The regression program calculates the **coefficient's own SE**, considering the scatter around the line and spread of X values.


### Calculate the standard error of the OLS slope **step by step**

Our fitted line for the **240 students** is $\widehat{GPA}_i=b_0+b_1\,exam\_total_i$. The **population slope** $\beta_1$ is fixed but unknown; the **sample slope** $b_1$ (also written $\hat\beta_1$) changes across random samples. We estimate the standard deviation of that sampling distribution with $SE(b_1)$.

For **simple OLS with an intercept and one predictor** (`exam_total`), the usual textbook SE is:

$$\boxed{SE(b_1)=\sqrt{\frac{s^2}{S_{xx}}}=\frac{s}{\sqrt{\sum_{i=1}^{n}(x_i-\bar x)^2}}},\qquad S_{xx}=\sum_{i=1}^{n}(x_i-\bar x)^2.$$

Follow the five steps below **using our actual sample**:

1. **Predict GPA for each student.** $\hat y_i=b_0+b_1x_i$.
2. **Find each residual.** $e_i=y_i-\hat y_i$ is the vertical distance from a student's observed GPA to the fitted line.
3. **Measure unexplained GPA variation.** Add up squared residuals, $SSE=\sum e_i^2$, then estimate error variance $s^2=\dfrac{SSE}{n-2}$. We divide by **$n-2$** because we estimated **two coefficients** (intercept and slope). Its square root, $s$, is the *residual standard error* in GPA units.
4. **Measure the spread of exam scores.** $S_{xx}=\sum(x_i-\bar x)^2$; a wider range of X values makes the slope easier to estimate precisely.
5. **Combine the two parts.** $SE(b_1)=\sqrt{\dfrac{SSE/(n-2)}{S_{xx}}}$. Larger residual scatter **increases** SE; wider X spread **decreases** SE, all else equal.

The regression package computes this for us; next we reproduce it **manually** and check that it matches `ols.bse["exam_total"]`. Note that this is **not** the sample-mean formula `SD / sqrt(n)`.


## Step R5. Why is there a p-value for a regression slope?

Ask a specific question about the **population**:

$$H_0: \beta_1=0\qquad H_1: \beta_1\ne0$$

- **$H_0$:** in the population, there is no *linear* association between exam total and GPA (zero slope).
- **$H_1$:** the population slope is not zero.

Follow these steps:

1. Fit a slope $b_1$ on **the sample**.
2. Calculate its **standard error**, $SE(b_1)$.
3. Compute **$t=b_1/SE(b_1)$**. This measures how many slope SEs the estimated slope is away from zero.
4. Calculate the **two-sided p-value** under the zero-slope model (both tails of a t distribution).
5. If $p<0.05$, reject $H_0$ at the chosen 5% level. Otherwise do not reject it.

## Step R8. What is the true slope for our INVENTED population?

We normally do **not** know this. Because the 3,000 students were created on a computer, we can fit the full-population relationship **only as a check AFTER fitting our sample model**.

Compare the sample slope and the full-population slope. They should be reasonably close, but they do not have to be identical.

# Logistic Regression (Logit): GPA + Exam Total → IELTS Pass (0/1)

## Why is this classification and not linear regression?

Here the outcome is not a GPA number. It is **yes/no**:

- **Y = `ielts_pass` = 1** if invented IELTS score **> 6.0**.
- **Y = `ielts_pass` = 0** if invented IELTS score **≤ 6.0**.

The predictors are **X = GPA and entrance-exam total score**. Do **not** put `ielts_score` into the predictors; it directly tells us the answer and would cause **data leakage**. We also do not include `ielts_pass_within_1year`.

**Logistic regression (Logit)** predicts a probability between **0% and 100%**. It is a regression *model* used for **binary classification**.


## Logit and predict a probability

Logit creates a score from the inputs, then converts that score into a predicted **probability**:

$$p=P(ielts\_pass=1\mid GPA,\ exam\_total)$$

If a student gets **$p=0.70$**, the model estimates a **70% probability** of having an IELTS score above 6.0 for students with those predictor values. We do **not** know which particular student will pass with certainty.

For each predictor, a **positive coefficient** means higher estimated pass *odds*, holding the other predictor constant; a **negative coefficient** means lower estimated odds.


## Step C4. Infer population relationships from the sample Logit coefficients

We want to learn about unknown **population** log-odds coefficients $\beta_{GPA}$ and $\beta_{exam}$, not just describe these 180 students.

For each predictor **holding the other predictor fixed**, we test:

$$H_0:\beta=0\qquad H_1:\beta\ne0$$

**Step by step:**

1. Estimate a coefficient **$b$** from our sample.
2. Find its **standard error, $SE(b)$**.
3. Calculate the **Wald Z-statistic: $z=b/SE(b)$**. This counts how many coefficient-SE units the estimated value is from zero.
4. Get a **two-sided p-value** from the normal distribution: $p=2P(Z\geq|z|)$.
5. Compare it with 0.05: a smaller p-value is evidence against a zero coefficient, *under model assumptions*.
6. Check a **95% confidence interval**. If it includes zero, the two-sided Wald p-value is generally above 0.05.


## Convert log-odds coefficients into odds ratios

Logit coefficients are hard to understand directly. We can calculate **odds ratio = $e^b$**.

- **OR = 1:** no difference in odds associated with one more unit of the predictor.
- **OR > 1:** higher odds of IELTS pass.
- **OR < 1:** lower odds of IELTS pass.


## Was our classification good? Check the TEST sample

Compare predictions with the invented outcomes of the 60 **test students**:

- **Accuracy:** percent of test students correctly classified.
- **Precision:** among students predicted **pass**, how many actually passed?
- **Recall:** among students who actually passed, how many did we predict as pass?
- **Confusion matrix:** a table of correct and incorrect predictions.
